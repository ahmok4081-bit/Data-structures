O(n^2)

```python
def print_items(n):
    for i in range(n):
        for j in range(n):
            print(i, j)

print_items(10)
```

drop none dominate

```python
def print_items(n):
  for i in range(n):
    for j in range(n):
      print(i , j)

  for k in range(n):
    print(k)
print_items(10)
```
