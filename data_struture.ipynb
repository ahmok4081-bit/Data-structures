```python
# Student Score Manager

scores = []  # dynamic array

def add_score():
    score = int(input("Enter score: "))
    scores.append(score)  # dynamic array grows automatically
    print("Score added!\n")

def view_scores():
    print("All Scores:", scores, "\n")

def average_score():
    if len(scores) == 0:
        print("No scores available\n")
        return
    avg = sum(scores) / len(scores)
    print("Average Score:", avg, "\n")

def highest_score():
    if len(scores) == 0:
        print("No scores available\n")
        return
    print("Highest Score:", max(scores), "\n")


# simple menu
while True:
    print("1. Add Score")
    print("2. View Scores")
    print("3. Average Score")
    print("4. Highest Score")
    print("5. Exit")

    choice = input("Choose: ")

    if choice == "1":
        add_score()
    elif choice == "2":
        view_scores()
    elif choice == "3":
        average_score()
    elif choice == "4":
        highest_score()
    elif choice == "5":
        break
    else:
        print("Invalid choice\n")
```
