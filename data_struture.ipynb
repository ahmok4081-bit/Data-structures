```python
# Student Score Manager

scores = []  # dynamic array

def add_score():
    score = int(input("Enter score: "))
    scores.append(score)  # dynamic array grows automatically
    print("Score added!\n")

def view_scores():
    print("All Scores:", scores, "\n")

def average_score():
    if len(scores) == 0:
        print("No scores available\n")
        return
    avg = sum(scores) / len(scores)
    print("Average Score:", avg, "\n")

def highest_score():
    if len(scores) == 0:
        print("No scores available\n")
        return
    print("Highest Score:", max(scores), "\n")


# simple menu
while True:
    print("1. Add Score")
    print("2. View Scores")
    print("3. Average Score")
    print("4. Highest Score")
    print("5. Exit")

    choice = input("Choose: ")

    if choice == "1":
        add_score()
    elif choice == "2":
        view_scores()
    elif choice == "3":
        average_score()
    elif choice == "4":
        highest_score()
    elif choice == "5":
        break
    else:
        print("Invalid choice\n")
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Create nodes
node1 = Node(15)
node2 = Node(3)
node3 = Node(17)
node4 = Node(90)

# Link nodes
node1.next = node2
node2.next = node3
node3.next = node4

head = node1  # Head points to the first node

# Traverse and print the linked list
current = head
while current:
    print(current.data, end=" -> ")
    current = current.next
print("None")
```

```python
class SinglyNode:
  def __init__(self, val, next= None):
    self.val = val
    self.next = next
  def __str__(self):
    return str(self.val)
```

```python
Head = SinglyNode(1)
A = SinglyNode(3)
B = SinglyNode(4)
c = SinglyNode(7)
Head.next = A
A.next = B
B.next = c

curr = Head
while curr:
  print(curr)
  curr = curr.next
```

```python
Head = SinglyNode(1)
A = SinglyNode(3)
B = SinglyNode(4)
c = SinglyNode(7)
Head.next = A
A.next = B
B.next = c

def search(Head, val):
  curr = Head # Corrected 'crr' to 'curr'
  while curr:
    if val == curr.val:
      return True
    curr = curr.next
  return False
search(Head, 9 )
```
